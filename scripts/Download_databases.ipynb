```python
import io
import zipfile
import requests
import gzip
```

```python
url = "https://coconut.s3.uni-jena.de/prod/downloads/2026-07/coconut_csv-07-2026.zip"
r = requests.get(url)
r.raise_for_status()

with zipfile.ZipFile(io.BytesIO(r.content)) as z:
    csv_name = next(
        name for name in z.namelist()
        if name.lower().endswith(".csv")
    )

    with z.open(csv_name) as source, open("coconut.csv", "wb") as destination:
        destination.write(source.read())
```

```python
chebi_urls = [ 
    "https://ftp.ebi.ac.uk/pub/databases/chebi/flat_files/compounds.tsv.gz"
]
for url in chebi_urls:
    print(f"Downloadding {url}")
    r = requests.get(url)
    r.raise_for_status()

with gzip.open(io.BytesIO(r.content), "rb") as source:
    with open("chebi.csv", "wb") as destination:
        destination.write(source.read())
```
